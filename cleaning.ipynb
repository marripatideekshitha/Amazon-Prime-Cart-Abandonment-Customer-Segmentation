```python
df["Is_Cart_Session"] = (df["Items_in_Cart"] > 0).astype(int)

df["Is_Abandoned"] = (
    (df["Items_in_Cart"] > 0) &
    (df["Drop_off_Stage"] != "Success")
).astype(int)
```

```python
def segment(row):
    if row["Items_in_Cart"] == 0:
        return "Browsers"
    if row["Drop_off_Stage"] == "Success":
        return "Purchasers"
    if row["Value_Flag"] == "Outlier" or pd.isna(row["Total_Cart_Value"]):
        return "Value Unknown/Outlier"
    if row["Total_Cart_Value"] < 50:
        return "Low-Value Abandoners"
    if row["Total_Cart_Value"] < 150:
        return "Mid-Value Abandoners"
    return "High-Value Abandoners"

df["Segment"] = df.apply(segment, axis=1)
df["Segment"].value_counts()
```
